* **Cells 2–5 — full re-analysis.** Re-estimates the complex-sample logistic regression from the KYRBS 2025 raw data file. The raw data are not redistributed here; they are obtained by application to the Korea Disease Control and Prevention Agency (KDCA). Authors (or readers who hold the data) can place the SPSS file in Google Drive and run these cells.

Design: stratification `STRATA`, clusters `CLUSTER`, weights `W`, with-replacement variance estimation (Taylor linearisation), design df = (clusters − strata) = 690. Group comparisons use subpopulation (domain) analysis; adolescents of undetermined multicultural status are excluded from the domains but retained in the design.

## 3. Complex-sample logistic regression (SPSS CSLOGISTIC equivalent)


```python
from scipy import stats
FACTORS=['SCHOOL','SEX','br2','ca','sw','ze','fa','ve','mi','ni','wa','fr','sleep','str_d','M_SAD','esrcrd_r','ese_r']
def dm(df):
    cols,names=[np.ones(len(df))],['Intercept']
    for f in FACTORS:
        lv=sorted(df[f].unique(),key=lambda x:(str(x) if f=='SCHOOL' else float(x)))
        for L in lv[:-1]: cols.append((df[f]==L).astype(float).values); names.append(f'{f}={L}')
    return np.column_stack(cols),names
def svy_logit(df,domain):
    X,names=dm(df); y=(df['s5']==1).astype(float).values
    w=np.where(domain,df['W'].values,0.0); strata,psu=df['STRATA'].values,df['CLUSTER'].values
    b=np.zeros(X.shape[1])
    for _ in range(50):
        p=1/(1+np.exp(-X@b)); s=np.linalg.solve(X.T@(X*(w*p*(1-p))[:,None]),X.T@(w*(y-p))); b+=s
        if np.max(np.abs(s))<1e-10: break
    p=1/(1+np.exp(-X@b)); H=X.T@(X*(w*p*(1-p))[:,None])
    U=pd.DataFrame(X*(w*(y-p))[:,None]); U['h']=strata; U['c']=psu
    Z=U.groupby(['h','c']).sum(); M=np.zeros((X.shape[1],)*2)
    for _,g in Z.groupby(level=0):
        k=len(g); A=g.values-g.values.mean(0)
        if k>1: M+=k/(k-1)*A.T@A
    Hi=np.linalg.inv(H); se=np.sqrt(np.diag(Hi@M@Hi))
    degf=pd.Series(psu).groupby(strata).nunique().sum()-df['STRATA'].nunique()
    tc=stats.t.ppf(0.975,degf); t=b/se
    return pd.DataFrame({'B':b,'lo':b-tc*se,'hi':b+tc*se,'p':2*stats.t.sf(np.abs(t),degf)},index=names),degf
main_NM,degf=svy_logit(d,d['multi3'].values==0)
main_MC,_  =svy_logit(d,d['multi3'].values==1)
sens_NM,_  =svy_logit(d,d['multi'].values==0)     # sensitivity: undetermined coded non-multicultural
print('design df:',degf)
```

## 4. Table 3 and comparison with the published values


```python
SPEC=[('Gender','Female (vs male)','SEX=1.0',1),
 ('School level','Middle school (vs vocational)','SCHOOL=중학교',-1),
 ('School level','General high (vs vocational)','SCHOOL=일반계고',-1),
 ('Economic status','Upper-middle (vs high)','ese_r=4.0',-1),('Economic status','Middle (vs high)','ese_r=3.0',-1),
 ('Economic status','Lower-middle (vs high)','ese_r=2.0',-1),('Economic status','Low (vs high)','ese_r=1.0',-1),
 ('Academic performance','Upper-middle (vs high)','esrcrd_r=4.0',-1),('Academic performance','Middle (vs high)','esrcrd_r=3.0',-1),
 ('Academic performance','Lower-middle (vs high)','esrcrd_r=2.0',-1),('Academic performance','Low (vs high)','esrcrd_r=1.0',-1),
 ('Dietary habits','Breakfast skipping (>=5 days/week)','br2=0.0',-1),('Dietary habits','Caffeinated beverage (high)','ca=0.0',1),
 ('Dietary habits','Sugar-sweetened beverage (high)','sw=0.0',1),('Dietary habits','Zero-calorie beverage (>=3/week)','ze=0.0',1),
 ('Dietary habits','Fast food (high)','fa=0.0',1),('Dietary habits','Vegetable (low)','ve=0.0',-1),('Dietary habits','Milk (low)','mi=0.0',-1),
 ('Dietary habits','Late-night snack (>=2 days/week)','ni=0.0',1),('Dietary habits','Water (low)','wa=0.0',-1),
 ('Dietary habits','Fruit (low)','fr=0.0',-1),('Sleep & mental health','Sleep insufficient','sleep=0.0',-1),
 ('Sleep & mental health','Stress (high)','str_d=0.0',1),('Sleep & mental health','Depression (yes)','M_SAD=0.0',1)]
def cell(r,k,s):
    B,lo,hi,p=r.loc[k,['B','lo','hi','p']]
    a,l,h=(np.exp(B),np.exp(lo),np.exp(hi)) if s==-1 else (np.exp(-B),np.exp(-hi),np.exp(-lo))
    return f'{a:.2f} ({l:.2f}-{h:.2f})',('<0.001' if p<0.001 else f'{p:.3f}')
out=[]
pub={lab:(nm,mc) for lab,nm,mc in [(e[0].strip(),e[1],e[2]) for e in EST if e[1]]}
for g,lab,k,s in SPEC:
    nm,np_=cell(main_NM,k,s); mc,mp=cell(main_MC,k,s); sn,sp=cell(sens_NM,k,s)
    out.append([g,lab,nm,np_,mc,mp,sn,sp])
T=pd.DataFrame(out,columns=['Group','Variable','NM aOR (95% CI)','NM p','MC aOR (95% CI)','MC p','Sensitivity NM aOR (95% CI)','p'])
pd.set_option('display.max_colwidth',42); pd.set_option('display.width',220)
T
```

## 5. Notes

* Reference categories: male; vocational high school; high economic status; high academic performance; and the healthier category of each lifestyle factor. Adjusted odds ratios above 1 therefore denote higher odds of prolonged smartphone use relative to the more advantaged or healthier group.
* Breakfast skipping follows the official KYRBS definition (not eating breakfast on ≥5 of the past 7 days).
* The sensitivity analysis reclassifies adolescents of undetermined multicultural status as non-multicultural, reproducing Supplementary Table S1.
